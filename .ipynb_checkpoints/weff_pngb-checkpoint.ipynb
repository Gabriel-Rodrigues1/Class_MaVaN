```python
import numpy as np
import matplotlib.pyplot as plt
from classy import Class
from getdist.mcsamples import loadMCSamples

# =========================
# 1. LOAD THE CHAIN
# =========================
chains = loadMCSamples("../chains_mavan_pplus/pngb_mavan_pplus")


if getattr(chains, "loglikes", None) is None:
    raise RuntimeError("The chains do not contain -log(likelihood) values.")

bestfit_idx = int(np.nanargmin(chains.loglikes))
bestfit_minus_loglike = float(chains.loglikes[bestfit_idx])
print(f"Best-fit chain row: {bestfit_idx}; -ln L = {bestfit_minus_loglike:.6f}")

# The best-fit curve is evaluated separately. These weighted posterior
# realizations are used only to estimate the marginalized credible bands.
nsamples = 100
rng = np.random.default_rng(12345)
weights = np.asarray(chains.weights, dtype=float)
weights = np.where(np.isfinite(weights) & (weights > 0.0), weights, 0.0)
if weights.sum() == 0.0:
    raise RuntimeError("The chain contains no positive finite weights.")

# Stratified weighted resampling reduces Monte Carlo noise in the
# pointwise bands without increasing the number of CLASS evaluations.
posterior_cdf = np.cumsum(weights / weights.sum())
stratified_u = (np.arange(nsamples) + rng.random(nsamples)) / nsamples
posterior_indices = np.searchsorted(
    posterior_cdf, stratified_u, side="right"
)
rng.shuffle(posterior_indices)

# =========================
# 2. RECONSTRUCTION SETTINGS
# =========================
z_min, z_max = 0.0, 10.0
z_grid = np.linspace(z_min, z_max, 200)

# Thermal reference distribution for three degenerate, constant-mass
# neutrinos. These are the CLASS defaults used by the supplied chain.
T_CMB_K = 2.7255
T_NCDM_OVER_T_CMB = 0.71611
K_B_EV_PER_K = 8.617333262e-5
T_NCDM0_EV = T_CMB_K * T_NCDM_OVER_T_CMB * K_B_EV_PER_K

# Fixed Gauss-Legendre quadrature for the Fermi-Dirac momentum integrals.
# q=40 is safely in the exponentially suppressed tail of 1/(exp(q)+1).
_gl_nodes, _gl_weights = np.polynomial.legendre.leggauss(160)
_q_max = 40.0
_q = 0.5 * _q_max * (_gl_nodes + 1.0)
_dq_weights = 0.5 * _q_max * _gl_weights
_fermi_dirac = 1.0 / (np.exp(_q) + 1.0)
_rho_kernel = _dq_weights * _q**2 * _fermi_dirac
_pressure_kernel = _dq_weights * _q**4 * _fermi_dirac


def parameter_at(index, name):
    return float(chains.samples[index, chains.index[name]])


def class_parameters(index, use_theta_s=False, earlier_start=False):
    """Reproduce the physical model used to generate the supplied chain."""
    p = lambda name: parameter_at(index, name)
    params = {
        "omega_b": p("omega_b"),
        "omega_cdm": p("omega_cdm"),
        # Use derived H0 for the fast path. If that reconstruction fails,
        # retry with the sampled acoustic scale used by the original chain.
        "A_s": p("A_s"),
        "n_s": p("n_s"),
        "tau_reio": p("tau_reio"),
        # Exact ncdm convention of the Cobaya input: one distribution with
        # degeneracy 3. Thus m_ncdm is the mass of each neutrino.
        "N_ncdm": 1,
        "deg_ncdm": 3,
        "N_ur": 0.00641,
        "T_ncdm": T_NCDM_OVER_T_CMB,
        "m_ncdm": p("m_ncdm"),
        "beta_ncdm": p("beta_ncdm"),
        "scf_parameters": f"{p('scf_lambda')}, -7.00, 0.40, 0.0",
        "Omega_Lambda": 0.0,
        "Omega_fld": 0.0,
        "Omega_scf": -1.0,
        "attractor_ic_scf": "no",
        "scf_tuning_index": 1,
    }
    if use_theta_s:
        params["theta_s_100"] = p("theta_s_100")
    else:
        params["H0"] = p("H0")
    if earlier_start:
        # Numerical fallback requested by CLASS when its default a_ini does
        # not yet satisfy the radiation-domination tolerance.
        params["a_ini_over_a_today_default"] = 1.0e-16
    return params


def class_background(index, use_theta_s=False, earlier_start=False):
    """Compute and copy only the background columns needed below."""
    cosmo = Class()
    try:
        cosmo.set(
            class_parameters(
                index,
                use_theta_s=use_theta_s,
                earlier_start=earlier_start,
            )
        )
        cosmo.compute()
        bg = cosmo.get_background()
        required = (
            "z",
            "(.)rho_scf",
            "(.)p_scf",
            "(.)rho_ncdm[0]",
            "(.)p_ncdm[0]",
        )
        return {name: np.asarray(bg[name]).copy() for name in required}
    finally:
        try:
            cosmo.struct_cleanup()
        except Exception:
            pass
        try:
            cosmo.empty()
        except Exception:
            pass


def constant_mass_reference(z, mass_eV, rho_nu0):
    """Return rho_bar_nu and p_bar_nu for a constant present-day mass.

    The overall CLASS density normalization is fixed by rho_nu(z=0), while
    the exact redshift dependence follows the Fermi-Dirac momentum integrals.
    Degeneracy cancels in the ratios because it is identical in the true and
    reference neutrino sectors.
    """
    z = np.asarray(z, dtype=float)
    y = mass_eV / (T_NCDM0_EV * (1.0 + z))
    epsilon = np.sqrt(_q[None, :] ** 2 + y[:, None] ** 2)
    integral_rho = np.sum(_rho_kernel[None, :] * epsilon, axis=1)
    integral_pressure = np.sum(_pressure_kernel[None, :] / epsilon, axis=1)

    y0 = mass_eV / T_NCDM0_EV
    epsilon0 = np.sqrt(_q**2 + y0**2)
    integral_rho0 = np.sum(_rho_kernel * epsilon0)

    redshift_factor = (1.0 + z) ** 4
    rho_reference = rho_nu0 * redshift_factor * integral_rho / integral_rho0
    p_reference = (
        rho_nu0
        * redshift_factor
        * integral_pressure
        / (3.0 * integral_rho0)
    )
    return rho_reference, p_reference


def reconstruct(index):
    """Return the full apparent EoS and reconstruction diagnostics."""
    used_earlier_start = False
    try:
        bg = class_background(index)
    except Exception as first_error:
        try:
            # Reproduce the sampled Cobaya input rather than relying on
            # the stored, rounded derived H0 for numerically sensitive rows.
            bg = class_background(index, use_theta_s=True)
            print(f"Recovered chain row {index} using sampled theta_s_100.")
        except Exception as theta_error:
            error_text = str(theta_error)
            if (
                "Omega_r" not in error_text
                and "tol_initial_Omega_r" not in error_text
            ):
                raise theta_error from first_error
            bg = class_background(
                index, use_theta_s=True, earlier_start=True
            )
            used_earlier_start = True

    order = np.argsort(bg["z"])
    z_sorted = bg["z"][order]

    def on_grid(name):
        return np.interp(z_grid, z_sorted, bg[name][order])

    rho_scf = on_grid("(.)rho_scf")
    p_scf = on_grid("(.)p_scf")
    rho_nu = on_grid("(.)rho_ncdm[0]")
    p_nu = on_grid("(.)p_ncdm[0]")

    # Intrinsic canonical scalar-field EoS.
    w_phi = np.divide(
        p_scf,
        rho_scf,
        out=np.full_like(rho_scf, np.nan, dtype=float),
        where=rho_scf != 0.0,
    )

    # Full apparent-EoS definition:
    #   rho_DE_app = rho_phi + rho_nu - rho_nu_bar,
    #   p_DE_app   = p_phi   + p_nu   - p_nu_bar.
    # Unlike the low-z approximation, this remains valid when the interacting
    # neutrinos become semi-relativistic.
    rho_nu_reference, p_nu_reference = constant_mass_reference(
        z_grid,
        parameter_at(index, "m_ncdm"),
        rho_nu[0],
    )
    rho_de_app = rho_scf + rho_nu - rho_nu_reference
    p_de_app = p_scf + p_nu - p_nu_reference
    w_app = np.divide(
        p_de_app,
        rho_de_app,
        out=np.full_like(rho_de_app, np.nan, dtype=float),
        where=rho_de_app > 0.0,
    )

    pressure_ratio = np.divide(
        3.0 * p_nu,
        rho_nu,
        out=np.zeros_like(p_nu),
        where=rho_nu != 0.0,
    )
    max_pressure_ratio = float(np.nanmax(np.abs(pressure_ratio)))

    # The true and reference sectors coincide at z=0. This also checks the
    # thermal-integral normalization against the CLASS background output.
    actual_w_nu0 = p_nu[0] / rho_nu[0]
    reference_w_nu0 = p_nu_reference[0] / rho_nu_reference[0]
    reference_w0_mismatch = float(abs(actual_w_nu0 - reference_w_nu0))

    return (
        w_app,
        w_phi,
        rho_de_app,
        max_pressure_ratio,
        reference_w0_mismatch,
        used_earlier_start,
    )


# =========================
# 3. BEST-FIT CURVE
# =========================
print(
    "Best-fit model: "
    f"sum_mnu={3.0 * parameter_at(bestfit_idx, 'm_ncdm'):.6g} eV, "
    f"beta={parameter_at(bestfit_idx, 'beta_ncdm'):.6g}, "
    f"f={parameter_at(bestfit_idx, 'scf_lambda'):.6g}, "
    f"H0={parameter_at(bestfit_idx, 'H0'):.6g}"
)

try:
    (
        bestfit_wapp,
        bestfit_wphi,
        bestfit_rhoapp,
        bestfit_pressure_ratio,
        bestfit_reference_mismatch,
        bestfit_used_earlier_start,
    ) = reconstruct(bestfit_idx)
except Exception as exc:
    raise RuntimeError(
        "The best-fit row could not be reconstructed with the exact chain model."
    ) from exc

# =========================
# 4. POSTERIOR CREDIBLE BANDS
# =========================
all_wapp = []
all_wphi = []
max_pressure_ratios = [bestfit_pressure_ratio]
reference_mismatches = [bestfit_reference_mismatch]
earlier_start_rows = set()
if bestfit_used_earlier_start:
    earlier_start_rows.add(bestfit_idx)

bestfit_result = (
    bestfit_wapp,
    bestfit_wphi,
    bestfit_rhoapp,
    bestfit_pressure_ratio,
    bestfit_reference_mismatch,
    bestfit_used_earlier_start,
)
cache = {bestfit_idx: bestfit_result}
failed_indices = {}

for draw_number, index in enumerate(posterior_indices, start=1):
    index = int(index)
    if index in cache:
        result = cache[index]
    elif index in failed_indices:
        result = None
    else:
        try:
            result = reconstruct(index)
            cache[index] = result
        except Exception as exc:
            failed_indices[index] = str(exc).strip()
            result = None
            if len(failed_indices) <= 3:
                short_error = " ".join(str(exc).strip().splitlines()[-2:])
                print(f"CLASS failed for posterior row {index}: {short_error}")

    if result is not None:
        (
            wapp_curve,
            wphi_curve,
            _,
            pressure_check,
            reference_mismatch,
            used_earlier_start,
        ) = result
        all_wapp.append(wapp_curve)
        all_wphi.append(wphi_curve)
        max_pressure_ratios.append(pressure_check)
        reference_mismatches.append(reference_mismatch)
        if used_earlier_start:
            earlier_start_rows.add(index)

    if draw_number % 10 == 0 or draw_number == nsamples:
        print(
            f"Posterior draws processed: {draw_number}/{nsamples}; "
            f"successful: {len(all_wapp)}"
        )

minimum_successful = max(20, int(0.8 * nsamples))
if len(all_wapp) < minimum_successful:
    raise RuntimeError(
        f"Only {len(all_wapp)}/{nsamples} posterior draws succeeded. "
        "Do not use the resulting bands; check that the reconstruction CLASS "
        "build is the same build used by Cobaya."
    )

all_wapp = np.asarray(all_wapp)
all_wphi = np.asarray(all_wphi)

low95_wapp, low68_wapp, median_wapp, high68_wapp, high95_wapp = np.nanpercentile(
    all_wapp, [2.5, 16.0, 50.0, 84.0, 97.5], axis=0
)
low95_wphi, low68_wphi, median_wphi, high68_wphi, high95_wphi = np.nanpercentile(
    all_wphi, [2.5, 16.0, 50.0, 84.0, 97.5], axis=0
)

if earlier_start_rows:
    print(
        f"Numerical early-start fallback used for {len(earlier_start_rows)} "
        "unique chain rows."
    )
if failed_indices:
    print(
        f"WARNING: {len(failed_indices)} unique posterior rows still failed "
        "after the early-start retry."
    )


def first_phantom_crossing(w_curve, rho_curve):
    y = w_curve + 1.0
    valid = np.isfinite(y) & (rho_curve > 0.0)
    crossing_idx = np.where(
        valid[:-1] & valid[1:] & (y[:-1] * y[1:] < 0.0)
    )[0]
    if not crossing_idx.size:
        return np.nan
    i = crossing_idx[0]
    z1, z2 = z_grid[i], z_grid[i + 1]
    w1, w2 = w_curve[i], w_curve[i + 1]
    return z1 + (-1.0 - w1) * (z2 - z1) / (w2 - w1)


bestfit_z_cross = first_phantom_crossing(bestfit_wapp, bestfit_rhoapp)
if np.isfinite(bestfit_z_cross):
    print(f"Best-fit apparent-EoS crossing: z = {bestfit_z_cross:.3f}")
else:
    print("The best-fit apparent EoS does not cross w=-1 in the plotted range.")

print(
    "Maximum 3 p_nu/rho_nu in plotted reconstructions: "
    f"{np.nanmax(max_pressure_ratios):.3e} "
    "(included through the full pressure-dependent definition)"
)
max_reference_mismatch = float(np.nanmax(reference_mismatches))
print(
    "Maximum z=0 reference-neutrino EoS mismatch: "
    f"{max_reference_mismatch:.3e}"
)
if max_reference_mismatch > 1.0e-5:
    print(
        "WARNING: the reference Fermi-Dirac integral does not reproduce the "
        "CLASS z=0 neutrino pressure accurately. Check T_ncdm and T_cmb."
    )
```

```python
# =========================
# 5. PLOT POSTERIOR MEDIANS, BEST FITS, AND CREDIBLE BANDS
# =========================
plt.figure(figsize=(8, 5))

plt.fill_between(
    z_grid,
    low95_wapp,
    high95_wapp,
    color="teal",
    alpha=0.15,
)
plt.fill_between(
    z_grid,
    low68_wapp,
    high68_wapp,
    color="teal",
    alpha=0.35,
)
plt.plot(
    z_grid,
    median_wapp,
    color="teal",
    lw=2,
    label=r"Posterior median: $w_{\rm app}$",
)
plt.plot(
    z_grid,
    bestfit_wapp,
    color="teal",
    lw=2,
    ls="--",
    label=r"Best fit: $w_{\rm app}$",
)

plt.fill_between(
    z_grid,
    low95_wphi,
    high95_wphi,
    color="darkorange",
    alpha=0.12,
)
plt.fill_between(
    z_grid,
    low68_wphi,
    high68_wphi,
    color="darkorange",
    alpha=0.30,
)
plt.plot(
    z_grid,
    median_wphi,
    color="darkorange",
    lw=2,
    label=r"Posterior median: $w_\phi$",
)
plt.plot(
    z_grid,
    bestfit_wphi,
    color="darkorange",
    lw=2,
    ls="--",
    label=r"Best fit: $w_\phi$",
)

if np.isfinite(bestfit_z_cross):
    plt.axvline(
        bestfit_z_cross,
        color="black",
        ls=":",
        lw=1.5,
        label=fr"Best-fit crossing: $z={bestfit_z_cross:.2f}$",
    )

plt.axhline(-1.0, ls="--", color="gray")
plt.xlabel(r"$z$", fontsize=16)
plt.ylabel(r"$w(z)$", fontsize=16)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlim(z_min, z_max)
plt.legend(fontsize=9, ncol=2)
plt.savefig("wphi_wapp_reconstruction.pdf", bbox_inches="tight")
plt.show()
```
