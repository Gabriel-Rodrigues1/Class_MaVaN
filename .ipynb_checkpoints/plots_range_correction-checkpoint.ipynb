```python
import numpy as np
from getdist import plots, loadMCSamples
from getdist.types import ResultTable

# ============================================================
# CHAINS
# ============================================================

DESY5_CHAIN_DIR = r"/home/usuario/research/chains_mavan_desy5_final"
DESY5_ROOT = "pngb_mavan_desy5_final"

PPLUS_CHAIN_DIR = r"/home/usuario/research/chains_mavan_pplus_final"
PPLUS_ROOT = "pngb_mavan_pplus_final"

TRIANGLE_PARAMS = [
    "H0",
    "beta_ncdm",
    "f_pngb",
    "Omega_m",
    "sigma8",
    "phi_ini",
]

TRIANGLE_OUTPUT = "pngb_mavan_desy5_pplus_triangle.pdf"


# ============================================================
# GETDIST SETTINGS
# ============================================================

analysis_settings = {
    # Smoothing
    "smooth_scale_1D": 5.0,
    "smooth_scale_2D": 5.0,

    # High-resolution KDE grids
    "fine_bins": 2048,
    "fine_bins_2D": 512,

    # Boundary corrections
    "boundary_correction_order": 1,
    "mult_bias_correction_order": 1,

    # 1 sigma and 2 sigma marginalized regions
    "contours": [0.68, 0.95],
}


# ============================================================
# LOAD CHAINS
# ============================================================

desy5 = loadMCSamples(
    DESY5_CHAIN_DIR + "/" + DESY5_ROOT,
    settings=analysis_settings,
)

pplus = loadMCSamples(
    PPLUS_CHAIN_DIR + "/" + PPLUS_ROOT,
    settings=analysis_settings,
)


# ============================================================
# HARD PRIOR BOUNDARIES
# ============================================================
#
# These are especially important for the derived parameters
# f_pngb and phi_ini.
#
# Original sampled priors:
#   H0         in [55, 80]
#   beta_ncdm  in [-10, 10]
#   log_f_pngb in [ln(0.1), ln(3)]
#   theta_ini  in [0, pi]
#
# Therefore:
#   f_pngb     in [0.1, 3]
#   phi_ini    in [0, 3*pi]
# ============================================================

hard_ranges = {
#    "f_pngb": [0.09, 3.1],
    "phi_ini": [0.0, 3.0 * np.pi],
}

desy5.setRanges(hard_ranges)
pplus.setRanges(hard_ranges)

# Recompute statistics after setting the boundaries
desy5.updateBaseStatistics()
pplus.updateBaseStatistics()


# ============================================================
# TRIANGLE PLOT
# ============================================================

g = plots.get_subplot_plotter(subplot_size=2.7)

g.settings.num_plot_contours = 2
g.settings.axes_fontsize = 14
g.settings.axes_labelsize = 18
g.settings.legend_fontsize = 18
g.settings.title_limit_fontsize = 14
g.settings.title_limit_labels = True
g.settings.linewidth = 1.7
g.settings.linewidth_contour = 1.0
g.settings.alpha_filled_add = 0.80
g.settings.solid_contour_palefactor = 0.55
g.settings.axis_tick_max_labels = 4
g.settings.figure_legend_frame = False
g.settings.tight_layout = True

g.triangle_plot(
    # Keep order consistent with labels/colors:
    # Pantheon+ = red
    # DES Y5    = blue
    [pplus, desy5],

    TRIANGLE_PARAMS,
    filled=True,

    # 68% marginalized constraint above diagonal panels
    title_limit=2,

    legend_labels=[
        r"CMB + DESI DR2 + Pantheon+",
        r"CMB + DESI DR2 + DES Y5-Dovekie",
    ],

    contour_colors=[
        "red",
        "blue",
    ],

    line_args=[
        {"color": "red", "lw": 1.7},
        {"color": "blue", "lw": 1.7},
    ],

    contour_args=[
        {"alpha": 0.80},
        {"alpha": 0.70},
    ],
)

g.export(TRIANGLE_OUTPUT)
print("Saved triangle plot:", TRIANGLE_OUTPUT)
```

```python
# ============================================================
# 2-SIGMA / 95% MARGINALIZED STATISTICS
# ============================================================

# Get marginalized statistics AFTER applying the hard ranges
desy5_stats = desy5.getMargeStats()
pplus_stats = pplus.getMargeStats()

print("\n" + "=" * 90)
print("95% MARGINALIZED CONSTRAINTS (approximately 2 sigma)")
print("=" * 90)

table_2sigma = ResultTable(
    ncol=1,
    results=[
        pplus_stats,
        desy5_stats,
    ],
    paramList=TRIANGLE_PARAMS,
    limit=2,   # second contour = 95%
    titles=[
        r"CMB + DESI DR2 + Pantheon+",
        r"CMB + DESI DR2 + DES Y5-Dovekie",
    ],
)

print(table_2sigma.tableTex())
desy5_stats = desy5.getMargeStats()
pplus_stats = pplus.getMargeStats()

print(
    ResultTable(
        ncol=1,
        results=[pplus_stats, desy5_stats],
        paramList=TRIANGLE_PARAMS,
        limit=2,
        titles=[
            r"CMB + DESI DR2 + Pantheon+",
            r"CMB + DESI DR2 + DES Y5-Dovekie",
        ],
    ).tableTex()
)
```
