```python
# import classy module
from classy import Class
# uncomment to get plots displayed in notebook
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

# create an instance of CLASS
model1 = Class()
model2 = Class()
model3 = Class()
model4 = Class()
```

```python
# Set the model from CLASS
beta1 = {
        'omega_b': 0.022,
        'omega_cdm': 0.12,
        'H0': 70.0,
        'A_s': 2.100549e-9,
        'n_s': 0.96,
        'N_ur': 0.0,
        'N_ncdm': 1,
        'm_ncdm': 0.314,
        'T_ncdm': 0.7137658555,
        'deg_ncdm': 3,
        'beta_ncdm': 0.0,
        'Omega_fld': 0.0,
        'Omega_Lambda': 0.0,
        'Omega_scf': -1,
        # lambda, shooting guess for log10(V0), phi_ini, phi_prime_ini
        'scf_parameters': '1.0, -7.0, 0.0, 0.0',
        'scf_tuning_index': 1,
        'attractor_ic_scf': 'no',
        'ncdm_fluid_approximation': 3,
        'gauge': 'synchronous',
        'output':'mPk, tCl, pCl, lCl,dTk',
        'l_max_scalars': 2500,
        # A small margin prevents interpolation failure at the last k.
        'P_k_max_h/Mpc': 3.15,
        'non_linear': 'none',
        'lensing': 'yes',
        
}

model1.set(beta1)
model1.compute()





# Extract needed quantities for model 1

bg1 = model1.get_background()
# to check the quantities you can get from background
print(bg1.keys())

#redshift and scale factor
z1 = bg1['z']
a1 = 1.0 / (1.0 + z1)

#scalar field quantities
phi_scf1 = bg1['phi_scf']
dphi_scf1 = bg1['phi\'_scf']
rho_scf1 = bg1['(.)rho_scf']
p_scf1 = bg1['(.)p_scf']
V_scf1 = bg1['V_scf']
ddV_scf1 = bg1['V\'\'_scf']
w_phi1 =  p_scf1 / rho_scf1
# phi0 is the CURRENT value (at z=0, which is the LAST element)
phi0_1 = phi_scf1[-1]
#Neutrinos quantities
rho_nu1 = bg1['(.)rho_ncdm[0]']
p_nu1 = bg1['(.)p_ncdm[0]']
w_nu1 = p_nu1 / rho_nu1

drho_da1 = np.gradient(rho_nu1, a1)
dp_da1   = np.gradient(p_nu1, a1)

ca2_nu1 = dp_da1 / drho_da1

# The hubble parameter evolution
H1 = bg1['H [1/Mpc]']

# Calculate critical density
rho_crit1 = 3 * H1**2 / (8 * np.pi)  # in CLASS units





# Get beta from your parameters
beta1_val = 0.0


# Compute w_eff (equation 19)

H_conf1 = a1* H1 # conformal Hubble: H = a * H_cosmic

w_eff1 = w_phi1 + (beta1_val * dphi_scf1 * (rho_nu1 - 3*p_nu1)) / (3 * H_conf1* rho_scf1)






beta2 = {
        'omega_b': 0.022,
        'omega_cdm': 0.12,
        'H0': 70.0,
        'A_s': 2.100549e-9,
        'n_s': 0.96,
        'N_ur': 0.0,
        'N_ncdm': 1,
        'm_ncdm': 0.314,
        'T_ncdm': 0.7137658555,
        'deg_ncdm': 3,
        'beta_ncdm': 1.0,
        'Omega_fld': 0.0,
        'Omega_Lambda': 0.0,
        'Omega_scf': -1,
        # lambda, shooting guess for log10(V0), phi_ini, phi_prime_ini
        'scf_parameters': '1.0, -7.0, 0.0, 0.0',
        'scf_tuning_index': 1,
        'attractor_ic_scf': 'no',
        'ncdm_fluid_approximation': 3,
        'gauge': 'synchronous',
        'output':'mPk, tCl, pCl, lCl,dTk',
        'l_max_scalars': 2500,
        # A small margin prevents interpolation failure at the last k.
        'P_k_max_h/Mpc': 3.15,
        'non_linear': 'none',
        'lensing': 'yes',
        
}
model2.set(beta2)
model2.compute()


# Extract needed quantities for model 2

bg2 = model2.get_background()
# to check the quantities you can get from background
print(bg2.keys())

#redshift and scale factor
z2 = bg2['z']
a2 = 1.0 / (1.0 + z2)

#scalar field quantities
phi_scf2 = bg2['phi_scf']
dphi_scf2 = bg2['phi\'_scf']
rho_scf2 = bg2['(.)rho_scf']
p_scf2 = bg2['(.)p_scf']
V_scf2 = bg2['V_scf']
ddV_scf2 = bg2['V\'\'_scf']
w_phi2 =  p_scf2 / rho_scf2
# phi0 is the CURRENT value (at z=0, which is the LAST element)
phi0_2 = phi_scf2[-1]
#Neutrinos quantities
rho_nu2 = bg2['(.)rho_ncdm[0]']
p_nu2 = bg2['(.)p_ncdm[0]']
w_nu2 = p_nu2 / rho_nu2

drho_da2 = np.gradient(rho_nu2, a2)
dp_da2   = np.gradient(p_nu2, a2)

ca2_nu2 = dp_da2 / drho_da2

# The hubble parameter evolution
H2 = bg2['H [1/Mpc]']

# Calculate critical density
rho_crit2 = 3 * H2**2 / (8 * np.pi)  # in CLASS units




# Get beta from your parameters
beta2_val = 1.0


# Compute w_eff (equation 19)

H_conf2 = a2* H2 # conformal Hubble: H = a * H_cosmic

w_eff2 = w_phi2 + (beta2_val * dphi_scf2 * (rho_nu2 - 3*p_nu2)) / (3 * H_conf2* rho_scf2)




# beta=-0.79 e lambda = 1.22
####################################################
#
# Cosmological parameters and other CLASS parameters
#
####################################################
beta3 = {# LambdaCDM parameters (Planck 18 + lensing + BAO bestfit)
        'omega_b': 0.022,
        'omega_cdm': 0.12,
        'H0': 70.0,
        'A_s': 2.100549e-9,
        'n_s': 0.96,
        'N_ur': 0.0,
        'N_ncdm': 1,
        'm_ncdm': 0.314,
        'T_ncdm': 0.7137658555,
        'deg_ncdm': 3,
        'beta_ncdm': -0.79,
        'Omega_fld': 0.0,
        'Omega_Lambda': 0.0,
        'Omega_scf': -1,
        # lambda, shooting guess for log10(V0), phi_ini, phi_prime_ini
        'scf_parameters': '1.0, -7.0, 0.0, 0.0',
        'scf_tuning_index': 1,
        'attractor_ic_scf': 'no',
        'ncdm_fluid_approximation': 3,
        'gauge': 'synchronous',
        'output':'mPk, tCl, pCl, lCl,dTk',
        'l_max_scalars': 2500,
        # A small margin prevents interpolation failure at the last k.
        'P_k_max_h/Mpc': 3.15,
        'non_linear': 'none',
        'lensing': 'yes',
        
}
model3.set(beta3)
model3.compute()


# Extract needed quantities for model 3

bg3 = model3.get_background()
# to check the quantities you can get from background
print(bg3.keys())

#redshift and scale factor
z3 = bg3['z']
a3 = 1.0 / (1.0 + z3)

#scalar field quantities
phi_scf3 = bg3['phi_scf']
dphi_scf3 = bg3['phi\'_scf']
rho_scf3 = bg3['(.)rho_scf']
p_scf3 = bg3['(.)p_scf']
V_scf3 = bg3['V_scf']
ddV_scf3 = bg3['V\'\'_scf']
w_phi3 =  p_scf3 / rho_scf3
# phi0 is the CURRENT value (at z=0, which is the LAST element)
phi0_3 = phi_scf3[-1]

#Neutrinos quantities
rho_nu3 = bg3['(.)rho_ncdm[0]']
p_nu3 = bg3['(.)p_ncdm[0]']
w_nu3 = p_nu3 / rho_nu3

drho_da3 = np.gradient(rho_nu3, a3)
dp_da3   = np.gradient(p_nu3, a3)

ca2_nu3 = dp_da3 / drho_da3

# The hubble parameter evolution
H3 = bg3['H [1/Mpc]']

# Calculate critical density
rho_crit3 = 3 * H3**2 / (8 * np.pi)  # in CLASS units




# Get beta from your parameters
beta3_val = -0.79


# Compute w_eff (equation 19)

H_conf3 = a3* H3 # conformal Hubble: H = a * H_cosmic

w_eff3 = w_phi3 + (beta3_val * dphi_scf3 * (rho_nu3 - 3*p_nu3)) / (3 * H_conf3* rho_scf3)



# beta=1 e lambda = 1.22
####################################################
#
# Cosmological parameters and other CLASS parameters
#
####################################################
beta4 = {# LambdaCDM parameters (Planck 18 + lensing + BAO bestfit)
        'omega_b': 0.022,
        'omega_cdm': 0.12,
        'H0': 70.0,
        'A_s': 2.100549e-9,
        'n_s': 0.96,
        'N_ur': 0.0,
        'N_ncdm': 1,
        'm_ncdm': 0.314,
        'T_ncdm': 0.7137658555,
        'deg_ncdm': 3,
        'beta_ncdm': 1.0,
        'Omega_fld': 0.0,
        'Omega_Lambda': 0.0,
        'Omega_scf': -1,
        # lambda, shooting guess for log10(V0), phi_ini, phi_prime_ini
        'scf_parameters': '0.5, -7.0, 0.0, 0.0',
        'scf_tuning_index': 1,
        'attractor_ic_scf': 'no',
        'ncdm_fluid_approximation': 3,
        'gauge': 'synchronous',
        'output':'mPk, tCl, pCl, lCl,dTk',
        'l_max_scalars': 2500,
        # A small margin prevents interpolation failure at the last k.
        'P_k_max_h/Mpc': 3.15,
        'non_linear': 'none',
        'lensing': 'yes',
}
model4.set(beta4)
model4.compute()



# Extract needed quantities for model 4

bg4 = model4.get_background()
# to check the quantities you can get from background
print(bg4.keys())

#redshift and scale factor
z4 = bg4['z']
a4 = 1.0 / (1.0 + z4)

#scalar field quantities
phi_scf4 = bg4['phi_scf']
dphi_scf4 = bg4['phi\'_scf']
rho_scf4 = bg4['(.)rho_scf']
p_scf4 = bg4['(.)p_scf']
V_scf4 = bg4['V_scf']
ddV_scf4 = bg4['V\'\'_scf']
w_phi4 =  p_scf4 / rho_scf4
# phi0 is the CURRENT value (at z=0, which is the LAST element)
phi0_4 = phi_scf4[-1]
#Neutrinos quantities
rho_nu4 = bg4['(.)rho_ncdm[0]']
p_nu4 = bg4['(.)p_ncdm[0]']
w_nu4 = p_nu4 / rho_nu4

drho_da4 = np.gradient(rho_nu4, a4)
dp_da4   = np.gradient(p_nu4, a4)

ca2_nu4 = dp_da4 / drho_da4

# The hubble parameter evolution
H4 = bg4['H [1/Mpc]']

# Calculate critical density
rho_crit4 = 3 * H4**2 / (8 * np.pi)  # in CLASS units



# Get beta from your parameters
beta4_val = 1.0


# Compute w_eff (equation 19)

H_conf4 = a4* H4 # conformal Hubble: H = a * H_cosmic

w_eff4 = w_phi4 + (beta4_val * dphi_scf4 * (rho_nu4 - 3*p_nu4)) / (3 * H_conf4* rho_scf4)


```

```python
# get lesned C_l spectra
cls1 = model1.lensed_cl(2500)
cls2 = model2.lensed_cl(2500)
cls3 = model3.lensed_cl(2500)
cls4 = model4.lensed_cl(2500)

# Model 1: β=0

model1.set({'temperature contributions': 'lisw'})
model1.compute()
cl_eisw1 = model1.raw_cl(2500)

# Model 2: β=1

model2.set({'temperature contributions': 'lisw'})
model2.compute()
cl_eisw2 = model2.raw_cl(2500)

# Model 3: β=-0.7

model3.set({'temperature contributions': 'lisw'})
model3.compute()
cl_eisw3 = model3.raw_cl(2500)

# get multipoles l
ll1 = cls1['ell'][2:]
# get CMB temperature spectrum C_l^TT
clTT1 = cls1['tt'][2:]
# get CMB polarisation spectrum C_l^EE
clEE1 = cls1['ee'][2:]


ll2 = cls2['ell'][2:]
# get CMB temperature spectrum C_l^TT
clTT2 = cls2['tt'][2:]
# get CMB polarisation spectrum C_l^EE
clEE2 = cls2['ee'][2:]

ll3 = cls3['ell'][2:]
# get CMB temperature spectrum C_l^TT
clTT3 = cls3['tt'][2:]
# get CMB polarisation spectrum C_l^EE
clEE3 = cls3['ee'][2:]

ll4 = cls4['ell'][2:]
# get CMB temperature spectrum C_l^TT
clTT4 = cls4['tt'][2:]
# get CMB polarisation spectrum C_l^EE
clEE4 = cls4['ee'][2:]

# get P(k) at redhsift z=0
kk = np.geomspace(1e-4,3.0,num=1000) # k in h/Mpc
h1 = model1.h() # get reduced Hubble for conversions to h/Mpc
h2 = model2.h() # get reduced Hubble for conversions to h/Mpc
h3 = model3.h() # get reduced Hubble for conversions to h/Mpc

h4 = model4.h() # get reduced Hubble for conversions to h/Mpc

# Note that kk*h is the array of k in 1/Mpc
Pk1 = model1.get_pk_all(kk*h1,0.)*h1**3 # get power spectrum, converted into (Mpc/h)^3 units, instead of the default Mpc^3 units
Pk2 = model2.get_pk_all(kk*h2,0.)*h2**3 # get power spectrum, converted into (Mpc/h)^3 units, instead of the default Mpc^3 units
Pk3 = model3.get_pk_all(kk*h3,0.)*h3**3 # get power spectrum, converted into (Mpc/h)^3 units, instead of the default Mpc^3 units
Pk4 = model4.get_pk_all(kk*h4,0.)*h4**3 # get power spectrum, converted into (Mpc/h)^3 units, instead of the default Mpc^3 units


# By default, this is the non-linear matter power spectrum; to get the linear one,
# pass the optional argument 'nonlinear=False' to .get_pk_all()
```

```python
# Present-day mass of each ncdm particle represented by this distribution
mnu0 = 0.314  # eV

beta_ncdm1 = 0.0
beta_ncdm2 = 1.0
beta_ncdm3 = -0.79
beta_ncdm4 = 1.0

# Since phi0_i = phi_scf_i[-1], all curves satisfy m_eff(z=0) = mnu0
m_eff1 = mnu0 * np.exp(beta_ncdm1 * (phi_scf1 - phi0_1))
m_eff2 = mnu0 * np.exp(beta_ncdm2 * (phi_scf2 - phi0_2))
m_eff3 = mnu0 * np.exp(beta_ncdm3 * (phi_scf3 - phi0_3))
m_eff4 = mnu0 * np.exp(beta_ncdm4 * (phi_scf4 - phi0_4))

fig, ax = plt.subplots(figsize=(10, 6), layout="constrained")

ax.semilogx(
    1.0 + z1, m_eff1,
    color="black", linewidth=2,
    label=r"$\beta=0,\ \lambda=1$",
)
ax.semilogx(
    1.0 + z2, m_eff2,
    color="blue", linewidth=2,
    label=r"$\beta=1,\ \lambda=1$",
)
ax.semilogx(
    1.0 + z3, m_eff3,
    color="green", linewidth=2,
    label=r"$\beta=-0.79,\ \lambda=1$",
)
ax.semilogx(
    1.0 + z4, m_eff4,
    color="red", linewidth=2,
    label=r"$\beta=1,\ \lambda=0.5$",
)

ax.set_xlabel(r"$1+z$", fontsize=18)
ax.set_ylabel(r"$m_{\nu,\mathrm{eff}}\;[\mathrm{eV}]$", fontsize=18)

ax.set_xlim(1.0, 1.0e8)
# Do not impose 0.04--0.08 when m_nu,0 = 0.314 eV
# ax.set_ylim(0.04, 0.08)

ax.tick_params(axis="both", which="major", labelsize=16)
ax.grid(True, which="both", alpha=0.3)
ax.legend(fontsize=14, frameon=False)

fig.savefig("m_eff_vs_z.pdf", bbox_inches="tight")
plt.show()
```

```python
# plot C_l^TT
T_CMB = 2.7255e6
plt.figure(figsize=(10, 6))

plt.xscale('log')
plt.yscale('linear')
plt.xlim(2, 1500)

plt.xlabel(r'$\ell$', fontsize=18)
plt.ylabel(r'$[\ell(\ell+1)/2\pi]\, C_\ell^{\mathrm{TT}}$', fontsize=18)

# Increase tick size
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

# Plot lines with consistent linewidth
plt.plot(ll1, (clTT1*ll1*(ll1+1))/(2.*np.pi)*T_CMB**2,
         color='black', linestyle='-', linewidth=2, label=r'$\beta=0, \ \lambda=1$')

plt.plot(ll2, (clTT2*ll2*(ll2+1))/(2.*np.pi)*T_CMB**2,
         color='b', linestyle='--', linewidth=2, label=r'$\beta=1, \ \lambda=1$')

plt.plot(ll3, (clTT3*ll3*(ll3+1))/(2.*np.pi)*T_CMB**2,
         color='g', linestyle=':', linewidth=2, label=r'$\beta=-0.79, \ \lambda=1$')

plt.plot(ll4, (clTT4*ll4*(ll4+1))/(2.*np.pi)*T_CMB**2,
         color='r', linestyle='-.', linewidth=2, label=r'$\beta=1, \ \lambda=0.5$')

plt.legend(fontsize=16)
plt.grid(True, alpha=0.3)

plt.savefig('mnu3_cltt_beta.pdf', bbox_inches='tight')
plt.show()
```

```python
# plot P(k)
plt.figure(figsize=(10, 6))

plt.xscale('log')
plt.yscale('log')
plt.xlim(1e-4, 1e1)
plt.ylim(1e1, 1e5)

plt.xlabel(r'$k \,\, [h/\mathrm{Mpc}]$', fontsize=18)
plt.ylabel(r'$P(k) \,\, [\mathrm{Mpc}/h]^3$', fontsize=18)

# Increase tick label sizes
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

# Plot with consistent line width
plt.plot(kk, Pk1, 'black', linewidth=2, label=r'$\beta=0, \ \lambda=1$')
plt.plot(kk, Pk2, 'b-', linewidth=2, label=r'$\beta=1, \ \lambda=1$')
plt.plot(kk, Pk3, 'g-', linewidth=2, label=r'$\beta=-0.79, \ \lambda=1$')
plt.plot(kk, Pk4, 'r-', linewidth=2, label=r'$\beta=1, \ \lambda=0.5$')

plt.legend(fontsize=16)
plt.grid(True, alpha=0.3)

plt.savefig('mnu3_pk_beta.pdf', bbox_inches='tight')
plt.show()
```

```python
# Plot w_φ vs redshift
fig, ax = plt.subplots(figsize=(10, 6))

ax.semilogx(1 + z1, w_phi1, 'black', linewidth=2, label=r'$\beta=0, \lambda=1$')
ax.semilogx(1 + z2, w_phi2, 'b-', linewidth=2, label=r'$\beta=1, \lambda=1$')
ax.semilogx(1 + z3, w_phi3, 'g-', linewidth=2, label=r'$\beta=-0.79, \lambda=1$')
ax.semilogx(1 + z4, w_phi4, 'r-', linewidth=2, label=r'$\beta=1, \lambda=1$')

ax.set_xlabel('1 + z', fontsize=18)
ax.set_ylabel(r'$w_\phi$', fontsize=18)

ax.set_xlim(1e0, 1e2)
ax.set_ylim(-1.8, 0.0)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=14)

ax.axhline(-1, color='red', linestyle='--', label='w = -1')
ax.axhline(-1/3, color='orange', linestyle='--', label='w = -1/3')

ax.legend(fontsize=16)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('w_phi_eos.pdf')
plt.show()

# Plot w_eff vs redshift
fig, ax = plt.subplots(figsize=(10, 6))

ax.semilogx(1 + z1, w_eff1, 'black', linewidth=2, label=r'$\beta=0, \lambda=1$')
ax.semilogx(1 + z2, w_eff2, 'b-', linewidth=2, label=r'$\beta=1, \lambda=1$')
ax.semilogx(1 + z3, w_eff3, 'g-', linewidth=2, label=r'$\beta=-0.79, \lambda=1$')
ax.semilogx(1 + z4, w_eff4, 'r-', linewidth=2, label=r'$\beta=1, \lambda=1$')


ax.set_xlabel('1 + z', fontsize=18)
ax.set_ylabel(r'$w_{\rm eff}$', fontsize=18)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=16)


ax.set_xlim(1e0, 1e2)
ax.set_ylim(-1.8, 0.0)

ax.axhline(-1, color='red', linestyle='--', linewidth=1.5,
           label=r'$w = -1$ (cosmological constant)')
ax.axhline(-1/3, color='orange', linestyle='--', linewidth=1.5,
           label=r'$w = -1/3$ (acceleration threshold)')

ax.legend(fontsize=14)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('w_eff_eos.pdf', bbox_inches='tight')
plt.show()
```
